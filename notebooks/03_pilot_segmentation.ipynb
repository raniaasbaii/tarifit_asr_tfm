```python
# Load the model on the GPU
model = whisper.load_model("small", device="cuda")

# Transcribe the audio
result = model.transcribe(
    AUDIO_PATH,
    task="transcribe",
    verbose=True
)
```

```python
import json
import os

OUTPUT_DIR = "data/processed/transcripts_raw"
os.makedirs(OUTPUT_DIR, exist_ok=True)

output_path = os.path.join(
    OUTPUT_DIR,
    "REC038_whisper_small.json"
)

with open(output_path, "w", encoding="utf-8") as f:
    json.dump(
        result,
        f,
        ensure_ascii=False,
        indent=2
    )

print("Saved to:", output_path)
```

```python
from pathlib import Path
import subprocess

audio_path = Path(
    "/content/drive/MyDrive/tarifit_asr_tfm/data/processed/wav_16khz_mono/REC038.wav"
)

segments_dir = Path(
    "/content/drive/MyDrive/tarifit_asr_tfm/data/processed/segments/REC038"
)

segments_dir.mkdir(parents=True, exist_ok=True)

for i, segment in enumerate(result["segments"], start=1):
    start = segment["start"]
    end = segment["end"]
    duration = end - start

    output_file = segments_dir / f"REC038_{i:03d}.wav"

    subprocess.run(
        [
            "ffmpeg",
            "-y",
            "-i", str(audio_path),
            "-ss", str(start),
            "-t", str(duration),
            "-ar", "16000",
            "-ac", "1",
            str(output_file)
        ],
        stdout=subprocess.DEVNULL,
        stderr=subprocess.DEVNULL,
        check=True
    )

print("Segments created:", len(result["segments"]))
print("Saved in:", segments_dir)
```

```python
import pandas as pd
from pathlib import Path

rows = []

for i, segment in enumerate(result["segments"], start=1):
    rows.append({
        "segment_id": f"REC038_{i:03d}",
        "recording_id": "REC038",
        "start_time": round(segment["start"], 2),
        "end_time": round(segment["end"], 2),
        "duration_seconds": round(segment["end"] - segment["start"], 2),
        "audio_path": f"data/processed/segments/REC038/REC038_{i:03d}.wav",
        "whisper_draft": segment["text"].strip(),
        "transcript_raw": "",
        "transcript_normalized": "",
        "transcription_confidence": "",
        "included": ""
    })

df = pd.DataFrame(rows)

csv_path = Path(
    "/content/drive/MyDrive/tarifit_asr_tfm/data/metadata/REC038_segments.csv"
)

df.to_csv(csv_path, index=False, encoding="utf-8")

print("Saved:", csv_path)
print("Rows:", len(df))
df.head()
```
